```python
%%writefile promedio.py
def calcular_promedio(n1,n2,n3,n4):
  return (n1+n2+n3+n4)/4
import statistics
def media(n1,n2,n3,n4):
  return statistics.median([n1,n2,n3,n4])
def moda(n1,n2,n3,n4):
  return statistics.mode([n1,n2,n3,n4])
```
