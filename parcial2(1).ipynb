```python
import numpy as np
import pandas as pd

np.random.seed(123)

prueba_A = np.random.normal(loc=7.2, scale=0.4, size=50)
prueba_B = np.random.gamma(shape=306, scale=0.0229, size=50)

df = pd.DataFrame({
    "A": prueba_A,
    "B": prueba_B
})
```

```python
df.describe()
```

interpretacion
La prueba A contiene 50 elementos\
La prueba A tiene un promedio de 7.205309\
La prueba A tiene desviacion estandar de 16.315524\
La prueba A el minimo valor es 143.000000\
La prueba A el 25% valor es 155.250000  (el 25% mas grade po decirlo asi teniendo al 100% como el valor mas grande )\
La prueba A el valor medio 166.000000\
La prueba A el 75% es 181.750000\
En la prueba A el maxiomo valor o mas grade es 	196.000000

```python
from scipy.stats import shapiro
shapiro(df["A"])
```

```python
from scipy.stats import shapiro
shapiro(df["B"])
```

```python
from scipy import stats
stats.kendalltau(df["A"], df["B"])
```
