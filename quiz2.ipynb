```python
usuarios = {
    "ANA": "1234",
    "JUAN": "abcd"
}

try:
    usuario = input("Usuario: ").upper()
    password = input("Contraseña: ")

    if usuario not in usuarios:
        raise KeyError("El usuario no existe")

    if usuarios[usuario] != password:
        raise ValueError("Contraseña incorrecta")

    else:
        print("Bienvenido,", usuario)

except KeyError as e:
    print("Error:", e)

except ValueError as e:
    print("Error:", e)
```

```python
productos = {
    "pan": 1.5,
    "leche": 2.0,
    "huevos": 3.0
}

print("Productos disponibles:", productos)
try:
  nombre = input("¿Qué producto quieres?: ").lower()
  if nombre not in productos:
    raise KeyError("producto no disponible")
  cantidad = int(input("Cantidad: "))
  if cantidad <= 0:
    raise ValueError("La cantidad debe ser mayor a 0")
  total = productos[nombre] * cantidad
  print("Total a pagar:", total)
except KeyError as e:
  print("producto no disponible:", e)
except ValueError as e:
  print("Error de valor:", e)
```

Introducción al manejo de expresiones regulares en
Python




Una expresión regular es una secuencia de caracteres que define un patrón de
búsqueda.\
Se utilizan para:\
• Buscar texto,\
• Validar información,\
• Extraer datos,\
• Reemplazar palabras,\
• Limpiar cadenas de texto.\
En Python se trabajan con el módulo re.\

```python
import re
```

```python
texto = "python es un lenguaje"
resultado= re.search("python", texto)
resultado
```

```python
texto= "tengo 24 perros y 3 GATOS"
resultado= re.findall(r"\d",texto)
resultado
```

nos dice los digitos que tiene nuestra cadena individalmente serparado por comas

```python
texto= "tengo 242 perros y 3 GATOS"
resultado= re.findall(r"\d+",texto)
resultado
```

si le ponemos un mas nos da el numero exapto y si hay mas lo separa por comas

```python
texto= "tengo 242 perros y 3 GATOS"
resultado= re.findall(r"\w+",texto)
resultado
```

También encontrar palabras completas (mediante \w+ ).

```python
import re
texto = "Me gusta java"
nuevo= re.sub("java","python",texto)
nuevo
```

cambia la cadena selecionada por otra elegida

```python
import re
texto= "Universidad"
resultado= re.findall(r"[aeiou]", texto)
print(resultado)
```

También encontrar letras específicas

```python
import re
texto= "Python Es Genial"
resultado= re.findall(r"[A-Z]" , texto)
print(resultado)
```

Se pueden establecer letras mayúsculas

```python
import re
texto= "Hola mundo Python"
resultado= re.findall(r"\s", texto)
print(resultado)
```

El argumento \s representa espacios

```python
import re
texto= "Tengo 15 libros y 200 hojas"
resultado= re.findall(r"\d+" , texto)
print(resultado)
```

```python
import re
texto= "rojo,verde,azul"
resultado= re.split(", ", texto)
print(resultado)
```

separa texto

```python
import re
texto = "Correos : ana@gmail.com y juan@hotmail.com"
resultado= re.findall(r"\w+@\w+\.\w+", texto)
print(resultado)
```

Hacer una búsqueda sencillas de correos electrónicos

```python
import re
telefono = "3104567890"
resultado= re.fullmatch(r"\d{10}", telefono)
print(resultado)
```

fullmatch valida toda la cadena. En este caso que se tengan 10 dígitos.
