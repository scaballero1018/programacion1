```python
def suma(z1, z2):
    resultado = z1 + z2
    return resultado

def multiplicacion(z1, z2):
    resultado = z1 * z2
    return resultado

def raiz_n(z, n):
    resultado = z ** (1/n)
    return resultado

def pedir_complejo():
    while True:
        try:
            print("Ingrese la parte real:")
            real = float(input())

            print("Ingrese la parte imaginaria:")
            imag = float(input())

            z = complex(real, imag)
            return z

        except ValueError:
            print("Error: debe ingresar solo números.")


opcion = 0

while opcion != 4:

    print("\nMENU DE OPCIONES")
    print("1. Sumar numeros complejos")
    print("2. Multiplicar numeros complejos")
    print("3. Raiz n-esima de un numero complejo")
    print("4. Salir(si desea terminar)")

    try:
        opcion = int(input("Seleccione una opcion: "))

        if opcion == 1:
            print("Ingrese el primer numero complejo:")
            z1 = pedir_complejo()
            print("Ingrese el segundo numero complejo:")
            z2 = pedir_complejo()
            s = suma(z1, z2)
            print("La suma es:", s)

        elif opcion == 2:
            print("Ingrese el primer numero complejo:")
            z1 = pedir_complejo()
            print("Ingrese el segundo numero complejo:")
            z2 = pedir_complejo()
            s = multiplicacion(z1, z2)
            print("La multiplicacion es:", s)

        elif opcion == 3:
            print("Ingrese el numero complejo:")
            z = pedir_complejo()

            print("Ingrese el valor de n(indice de la raiz):")
            while True:
                try:
                    n = float(input())
                    if n <= 0:
                        print("debe ingresar un numero entero positivo")
                        continue
                    s = raiz_n(z, n)
                    print("La raiz es:", s)
                    break

                except ValueError:
                    print("Error: debe ingresar un numero entero positivo.")

        elif opcion == 4:
            print("Fin del programa")

        else:
            print("Opcion no valida")

    except ValueError:
        print("Error: debe ingresar un numero.")
```
