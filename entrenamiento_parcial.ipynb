```python
import numpy as np
import pandas as pd
np.random.seed(123)#para guardar la semilla
```

```python
datos= np.random.normal(loc=4.0, scale=0.5,size=50)#simula loc= ala media scale=es la desviacion estandar 0.5
df=pd.DataFrame({
    "nota":datos
})
```

```python
df.describe()
```

std es la desvidacion estandar

```python
media=df["nota"].mean()
```

```python
import matplotlib.pyplot as plt

df["nota"].hist()
plt.title("Histogram of Notas")
plt.xlabel("Nota Value")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm
import numpy as np

# Get the data
notas = df["nota"]

# Create a figure and axes
plt.figure(figsize=(10, 6))

# Plot histogram and KDE using seaborn
sns.histplot(notas, kde=True, stat="density", color="skyblue", label="Histograma y KDE")

# Fit a normal distribution to the data
mu, std = norm.fit(notas)

# Plot the PDF of the fitted normal distribution
x = np.linspace(notas.min(), notas.max(), 100)
p = norm.pdf(x, mu, std)
plt.plot(x, p, 'r--', linewidth=2, label=f'Ajuste Normal ($\mu$={mu:.2f}, $\sigma$={std:.2f})')

# Add titles and labels for a 'bonito' plot
plt.title('Distribución de Notas con KDE y Ajuste Normal', fontsize=16)
plt.xlabel('Valor de Nota', fontsize=12)
plt.ylabel('Densidad', fontsize=12)
plt.legend(fontsize=10)
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
from scipy.stats import shapiro
shapiro(df["nota"])
```

como el valor p es mayor que 0.05, no se rechaza la hipotesis de la  normalidad

si fuera menor que 0.05 se rechaza la hipotesis de normalidad

```python
import numpy as np
import pandas as pd

np.random.seed(123)

grupo_A = np.random.normal(loc=4.0, scale=0.4, size=40)
grupo_B = np.random.normal(loc=4.3, scale=0.5, size=40)

df = pd.DataFrame({
    "A": grupo_A,
    "B": grupo_B
})

```

```python
from scipy import stats
t_stat, p_val = stats.ttest_ind(df["A"], df["B"])

print("Estadístico t:", t_stat)
print("Valor p:", p_val)
```

como los valores de p son menores de 0.05 se rechaza la hipotesis de la normal con una confianza del 95%
