```python
import re

class CorreoInvalidoError(Exception):
    pass

def validar_correo(correo):

    patron = r"^[a-z]+_[0-9]{2}_[a-z]+@universitaria\.edu\.co$"    #como se puede ver cambie esta parte del codigo de ayuda porque se podia tener una edad de 2718

    if not re.fullmatch(patron, correo):
        raise CorreoInvalidoError(
            "Correo inválido"
        )

    return True


intentos = 0

while intentos < 3:

    try:
        correo = input(
            "Ingrese su correo institucional: "
        )

        validar_correo(correo)

        print("Correo válido")
        print("Acceso permitido")

        break

    except CorreoInvalidoError as e:

        intentos += 1

        print("Error:", e)
        print(
            "Intento",
            intentos,
            "de 3"
        )

    except Exception as e:
        print(
            "Error inesperado:",
            e
        )

if intentos == 3:
    print(
        "Acceso bloqueado por intentos fallidos."
    )
```
